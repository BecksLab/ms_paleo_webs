### PERMANOVA Variance Partitioning

To quantify the relative contributions of model and temporal turnover to variation in inferred network structure, we conducted permutational multivariate analysis of variance (PERMANOVA). Euclidean distance matrices were calculated from standardised (z-transformed) network metrics. Model and extinction phase (‘time’) were analysed separately to estimate their total contributions to variance, and in combination to assess interaction effects. Significance was assessed using 999 permutations.

#### Robustness of model effects after temporal centering

To determine whether the dominance of model reflected absolute structural shifts among extinction phases, we repeated the analysis after centering network metrics within each time bin. This procedure removes mean temporal differences while preserving within-phase structural variation. Even after temporal centering, model type explained 84.8% of multivariate variance (R² = 0.848, p \< 0.001), exceeding the variance explained in the uncentered analysis. Thus, the strong influence of model identity is not attributable to temporal mean differences, but reflects intrinsic structural divergence among models.

## Statistical Drivers of Network Variation

### Statistical Robustness and Assumptions

Factorial ANOVA assumptions were validated via residual analysis. Despite significant heteroscedasticity (Levene’s test, p\<0.001), the perfectly balanced design (n = 100 per cell) and large sample size (N = 2400) ensure the robustness of the F-test. Visual inspection of Q-Q plots and Residuals-vs-Fitted plots confirmed that the distributions were sufficiently symmetric for parametric analysis.

## Constraining connectance decreases observed differences between models

It has been argued that connectance constrains emerging network properties ([Poisot and Gravel 2014](#ref-poisot2014); [Strydom et al. 2021](#ref-strydom2021)). To assess if constraining the connectance of *all* networks would result in similar network structures we actively constrained the connectance of all networks by randomly downsampling them to the target connectance *i.e.,* the PFIM, ATN, ADBM, and Body-size ratio models had links randomly prunes until they reached the target connectance. Note that in some instances the initial connectance of the network may have already been below the target connectance and in these instances the network was used ‘as is’. We repeated both the multivariate and temporal trajectory analyses for these connectance constrained networks.

### Figure S5. Constraining connectance decreases observed differences in network structure

Although there is still a significant effect of model type on network structure (Pillai’s trace = 1.64, approximate $F_{30,10080}$ = 164.5, $p<$ 0.001) we can see that the visual clustering is much less dramatic than for the unconstrained networks (Figure 1 in the main text). However we do not see connectance as the *main* driver of differences in network structure in the mulitvariate space – note the connectance is more strongly correlated with LDA axis 2 than with LDA axis 1 indicating that although it ‘reduces’ the differences in network structure between models it is not the main driver in the observed differences.



### Figure S6. Connectance increases the signal of time

When we look at the temporal changes in network structure there is qualitatively more agreement with the changes in network structure over time than for the unconstrained networks (Figure 2A in the main text) *i.e.,* we often see most models indicating similar changes in structure over time, although the magnitude of these changes remain different. Note that here we made no assumptions as to the expected changes in connectance over the extinction time and so there is a ‘random’ connectance signal across time - thus we can assume that the signal of temporal changes in network structure here still retain some sort of time $\times$ model signal. Indeed when we look at the two-way factorial ANOVA (panel B) we can see that model type is still the dominant determinant of network structure, however we see a greater source of variance coming from time (notably for vulnerability time becomes the dominant driver of variance).



Overall these results suggest that the specification of network connectance will impact the resulting overall network structure, however we still get distinct model-specific signature. This result is two-fold. Firstly it showcases that for any model where connectance is one of the parameters in the model there is a greater risk of enforcing differences or similarities in network structure depending on how connectance is specified – *e.g.,* in this instance any assumptions we make on expected change in connectance over time will drive at least in part any inferences we make about changes in network structure for both Niche and Random networks. Secondly, despite connectance-driven constraints we still see model-specific differences in structure which re-enforces that these models are still encoding fundamentally different processes that determine where links are present within the network.
